```python
# Ejercicio 2 — Series, Selección Condicional y Data Wrangling
import pandas as pd
import numpy as np
# Cargamos la nueva base de empleados (ya viene limpia)
df = pd.read_csv("empleados_technova_2024.csv")

df.info()   # 180 filas, 12 columnas, sin nulos -> no hay nada que limpiar
```

```python
df.head()  # Vemos las primeras filas para entender la estructura de los datos
```

```python
# Series en Pandas
# Crear una serie
mini_serie = pd.Series(
    [85000, 1200000, 950000],
    index=["Ana", "Carlos", "María"],
    name="salarios"
)

mini_serie
```

```python
mini_serie.to_frame()  # Convertir la serie a un DataFrame
```

```python
print("values: ", mini_serie.values)  # Valores de la serie
print("index: ", list(mini_serie.index))    # Índice de la serie
print("name: ", mini_serie.name)      # Nombre de la serie
print("dtype: ", mini_serie.dtype)    # Tipo de datos de la serie
```

```python
columna_salario = df["salario"]  # Seleccionar la columna "salario" como una serie
print(type(columna_salario))  # Verificar que es una serie
columna_salario.head()  # Ver las primeras filas de la serie de salarios
```

```python
# El guion bajo separador visual de miles
print(1_000_000 == 1000000)
print(1_000_000)

salario_sin = 12500000
salario_con = 12_500_000
print(salario_sin == salario_con)
```

```python
# Las series vectorizadas
(df["salario"] / 1_000_000).head()  # Dividir la serie de salarios por un millón para convertir a millones
```

```python
print("promedio: ", df["salario"].mean())  # Calcular el salario promedio
print("mediana: ", df["salario"].median())  # Calcular la mediana de los salarios
print("máximo: ", df["salario"].max())  # Calcular el salario máximo
print("mínimo: ", df["salario"].min())  # Calcular el salario mínimo

# idmax() devolve el índice del salario máximo
fila_max = df["salario"].idxmax()
print("índice del salario máximo: ", fila_max)  # Imprimir el índice del salario máximo

# con .loc obtiene ese indice para ver quien es esa persona
df.loc[fila_max, ["nombre", "departamento", "salario"]]  # Ver la fila completa del empleado con el salario máximo
```

```python
# Métodos de una serie para explorar valores

#* Valores distintos
print(df["departamento"].unique())  # Mostrar los departamentos únicos en la columna "departamento"

#* Cuenta cuantas veces aparece cada valor 
print(df["departamento"].value_counts())  # Contar cuántas veces aparece cada departamento en la columna "departamento"

print(df["departamento"].value_counts().sort_values())
```
