```python
# Operadores logicos, & (and), | (or), ~ (not)
# & (and) -> Y (ambas condiciones deben ser verdaderas)

# Personas del departamento de ventas en Santiago que ganan más de 1 millón
df[(df["departamento"] == "Ventas") & (df["ciudad"] == "Santiago") & (df["salario"] > 1_000_000)]
```

```python
# | (or) -> O (al menos una de las condiciones debe ser verdadera)
# Personas del departamento de Ventas o del departamento de Marketing
df[(df["departamento"] == "Ventas") | (df["departamento"] == "Marketing")]
```

```python
## ~ (not) -> NO (invierte la condición)

# Personas que No viven en Santiago
# ~ invierte la conndicion == 
fuera_santiago = df[~ (df["ciudad"] == "Santiago")]
print("personas que no viven en Santiago: ", len(fuera_santiago))
```

```python
# Obtener los que trabajan Remoto o Hibrido, pero que NO sean de gerencia
(df["departamento"] == "Gerencia")
```

```python
~(df["departamento"] == "Gerencia") # con el ~ se invierte la condición, es decir, selecciona a los que NO son de gerencia
```

```python
(df["modalidad"] == "Remoto") | (df["modalidad"] == "Híbrido")
```

```python
# Obtener los que trabajan Remoto o Hibrido, pero que NO sean de gerencia
flexibles = df[(df["modalidad"] == "Remoto") | (df["modalidad"] == "Híbrido") & ~(df["departamento"] == "Gerencia")]
print("personas que trabajan remoto o híbrido pero no son de gerencia: ", len(flexibles))
flexibles[["nombre", "departamento", "modalidad"]]
```
