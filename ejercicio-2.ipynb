```python
# Operadores logicos, & (and), | (or), ~ (not)
# & (and) -> Y (ambas condiciones deben ser verdaderas)

# Personas del departamento de ventas en Santiago que ganan más de 1 millón
df[(df["departamento"] == "Ventas") & (df["ciudad"] == "Santiago") & (df["salario"] > 1_000_000)]
```

```python
# | (or) -> O (al menos una de las condiciones debe ser verdadera)
# Personas del departamento de Ventas o del departamento de Marketing
df[(df["departamento"] == "Ventas") | (df["departamento"] == "Marketing")]
```

```python
## ~ (not) -> NO (invierte la condición)

# Personas que No viven en Santiago
# ~ invierte la conndicion == 
fuera_santiago = df[~ (df["ciudad"] == "Santiago")]
print("personas que no viven en Santiago: ", len(fuera_santiago))
```

```python
# Obtener los que trabajan Remoto o Hibrido, pero que NO sean de gerencia
(df["departamento"] == "Gerencia")
```

```python
~(df["departamento"] == "Gerencia") # con el ~ se invierte la condición, es decir, selecciona a los que NO son de gerencia
```

```python
(df["modalidad"] == "Remoto") | (df["modalidad"] == "Híbrido")
```

```python
# Obtener los que trabajan Remoto o Hibrido, pero que NO sean de gerencia
flexibles = df[(df["modalidad"] == "Remoto") | (df["modalidad"] == "Híbrido") & ~(df["departamento"] == "Gerencia")]
print("personas que trabajan remoto o híbrido pero no son de gerencia: ", len(flexibles))
flexibles[["nombre", "departamento", "modalidad"]]
```

```python
ventas_o_marketing = df.query('departamento == "Ventas" or departamento == "Marketing"') 
print(f"Personas en Ventas o Marketing: {len(ventas_o_marketing)} de {len(df)} " f"({len(ventas_o_marketing)/len(df):.1%} del total)")

print("\nReparto por departamento: ")
print(ventas_o_marketing["departamento"].value_counts())

# Renderizacion
display(ventas_o_marketing[["nombre", "departamento", "salario"]].head())
```

```python
# Querys acepta las sintaxis "in [lista]"
en_ciudades = df.query("ciudad in ['Santiago', 'Valparaíso', 'Concepción']")
print(f"Filas que cumplen: {len(en_ciudades)} de {len(df)} ({len(en_ciudades)/len(df):.1%} del total)")
print("\nCuántas personas hay en cada una de esas ciudades: ")
print(en_ciudades["ciudad"].value_counts())
display(en_ciudades[["nombre", "ciudad", "departamento"]].head())
```

```python
# "not in" es el inverso: conserva las filas cuyo valor No está en la lista
# Equivalente a: df[~df["modalidad"].isin(["Presencial"])]
no_presencial = df.query("modalidad not in ['Presencial']")
print(f"Personas que no trabajan de forma presencial: {len(no_presencial)} de {len(df)} ({len(no_presencial)/len(df):.1%} del total)")
print(no_presencial["modalidad"].value_counts())
display(no_presencial[["nombre", "modalidad"]].head())
```

```python
# Referencia a varialbes externas en query
umbral_salario = 1_800_000
umbral_evaluacion = 4.0

# query
top = df.query("salario > @umbral_salario and evaluacion_desempeno >= @umbral_evaluacion")
top[["nombre", "salario", "evaluacion_desempeno"]].head()

print(f"Cumplen salario > {umbral_salario} y evaluación >= {umbral_evaluacion}: {len(top)} personas")

# En python 0 se evalua como False
# Cualquier numero distinto de 0 se evalua como True
if len(top):
    print(f"Salario promedio del grupo: {top['salario'].mean():,.0f}")
    print(f"Evaluación promedio del grupo: {top['evaluacion_desempeno'].mean():.2f}")
# El if es una protección para no calcular promedios sobre un grupo vacio
display(top[["nombre", "salario", "evaluacion_desempeno"]].head())
```

```python
# @ con una lista 
departamentos_incluidos = ["TI", "Soporte", "Gerencia"]
seleccion = df.query("departamento in @departamentos_incluidos")
# seleccion[["nombre", "departamento"]]
print(seleccion.groupby("departamento").size())
print(seleccion.groupby("departamento")["salario"].mean().round(0))
```

```python
# Condición Compleja y Encadenamiento
# Query : Mujeres en modalidad Remoto o Híbrido con salario en un rango medio-alto y ordenadas por evaluación de mayor a menor
# un rango medio alto es aquel 1_000_000 <= salario <= 2_000_000

resultado = df.query("genero == 'Femenino' and modalidad in ['Remoto', 'Híbrido'] and salario.between(1_000_000, 2_000_000)"
).sort_values("evaluacion_desempeno", ascending=False)

print(f"Mujeres en Remoto/Híbrido con salario entre 1M y 2M: {len(resultado)} de {len(df)} ({len(resultado)/len(df):.1%} del total)")
```
